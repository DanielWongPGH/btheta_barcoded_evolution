```python
gene_array = np.array(sorted([[gene, start, stop] for gene, (start, stop, strand) in shared.gene_coords_map.items()]))
def find_gene_from_position(pos):
    gene_starts = gene_array[:,1].astype(int)
    gene_stops = gene_array[:,2].astype(int)
    gene_names = gene_array[:,0]

    start_bool = pos >= gene_starts
    stop_bool = pos <= gene_stops

    if (start_bool * stop_bool).sum() > 0:
        return gene_names[start_bool * stop_bool][0]
    
    else:
        upstream_gene = gene_names[start_bool][-1]
        downstream_gene = gene_names[stop_bool][0]
        return f'{upstream_gene}/{downstream_gene}'
    
def check_nearby(pos, arr_of_positions, dist = 1000):
    return np.any(np.abs(pos - arr_of_positions) <= dist)
```

```python
EXCLUDE_GENES = ['BT1040/BT1041', 'BT1042', 'BT3239/BT3240', 'BT4054/BT4055', 'BT2672/BT2673']
MOB_GENES = [f'BT{e}' for e in range(3134,3155)]
FIG4_SNP_GENES = ['BT0317', 'BT0867', 'BT1046', 'BT1573', 'BT3082', 'BT3465', 'BT4245', 'BT4247', 'BT4295', 'BT0900']
PLOT_GENES = ['BT0867', 'BT3465', 'BT2660', 'BT2661',
              'BT4014', 'BT4015',  'BT4540', 'BT4543', 'BT4520', 'BT4523'] + MOB_GENES + FIG4_SNP_GENES
PLOT_GENE_LOCATIONS = np.array([[shared.gene_coords_map[gene][0],shared.gene_coords_map[gene][1]] for gene in PLOT_GENES])
## we will only save timecourses for genes we plot


def load_mgx_timecourse(f):
    muts = {}
    with open(f) as f:
        for line in f:
            line_items = line.strip('\n').split(',')
            genome, pos, mut, days_str, reads_str, cov_str = line_items
            
            pos = int(pos)
            if not np.any( np.abs(PLOT_GENE_LOCATIONS - pos) <= 10000): 
                continue


            mut = mut.strip(' ')
            gene = find_gene_from_position(pos)
            if gene in EXCLUDE_GENES:
                continue
            
            days = np.array([float(x) for x in days_str.split(' ') if x != ''], dtype='int')
            reads = np.array([float(x) for x in reads_str.strip(' ').split(' ')])
            cov = np.array([float(x) for x in cov_str.strip(' ').split(' ')])

            muts[pos] = (mut, reads, cov)
            if gene not in muts:
                muts[gene] = []
            muts[gene].append( (pos, mut, reads, cov) )

            if 'junction' in mut:
                try:
                    mut = tuple([int(e) for e in mut.split('_')[1:]])
                    muts[mut] = (pos, reads, cov)
                except:
                    continue
            muts['days'] = days

            

    return muts


all_mutation_timecourses = {}
for file in sorted(os.listdir(mgx_timecourse_dir)):
    if file.endswith('merged_timecourse') and file.startswith('E'):
        file_metadata = file.split('_')
        if 'E1' in file:
            expt = 'E1'
        else: #E2
            expt = 'E2'
        mouse = int(file_metadata[1].strip('m'))

        print('Processing mouse', expt, mouse)
        all_mutation_timecourses[(expt, mouse)] = load_mgx_timecourse(f'{mgx_timecourse_dir}/{file}')
```

```python
def get_coverage_timecourses(file, span_dictionary):
    # span_dictionary = {'BT0867': (1066079, 1066179),
    #                    'BT3465': (4464553, 4464753),
    local_coverage_map = {key:{'single_site_coverages':[], 'genome_span':span} for key,span in span_dictionary.items()}
    
    with open(file) as f:
        header = next(f) #CHROM POS sample1 sample2
        header_items = header.strip('\n').split('\t')
        header_timepoint_map = {}

        sampled_times = []
        for i, sample in enumerate(header_items[2:]):
            sample_time = sample.split('_')[3]
            if 'cecum' in sample_time:
                sample_time = 100
            elif 'smallint' in sample_time:
                sample_time = 200
            else: 
                sample_time = int(sample_time.replace('day',''))
            header_timepoint_map[sample_time] = i+2 # column index
            sampled_times.append(sample_time)
        sampled_times = np.array(sorted(sampled_times))
        sorted_cols = np.array([header_timepoint_map[time] for time in sampled_times])

        
        for line in f:
            line_items = line.strip('\n').split('\t')
            pos = int(line_items[1])
            for key, (start, stop) in span_dictionary.items():
                if pos in range(start, stop+1):
                    sorted_sample_coverages = np.array([float(line_items[col]) for col in sorted_cols])
                    local_coverage_map[key]['single_site_coverages'].append(sorted_sample_coverages)
        
        for key in local_coverage_map:
            local_coverage_map[key]['times'] = sampled_times
            local_coverage_map[key]['average_coverage'] = np.mean(np.array(local_coverage_map[key]['single_site_coverages']), axis=0)
            del local_coverage_map[key]['single_site_coverages']
    return local_coverage_map
            
span_dictionary = {
#   'BT0867': (1066054, 1066104),
                   'BT0867': (1066079, 1066079),
                   'BT3465': (1065108, 1068370),
                   'BT3134-BT3154': (4001887-25, 4017683+25),  
                   'BT4014/BT4015': (5233175-25, 5233371+25),
                   'BT2660/BT2661': (3316290-25, 3316499+25),
                    'BT4540/BT4541': (5961716-25, 5962851+25),
                    'BT4540/BT4543': (5961162-25, 5964550+25),
                    'BT4520/BT4523': (5939745-25, 5943532+25),}
                    

all_coverage_timecourses = {}
for file in sorted(os.listdir(mgx_timecourse_dir)):
    if file.endswith('coverage_timecourse.txt') and file.startswith('E'):
        file_metadata = file.split('_')
        if 'E1' in file:
            expt = 'E1'
        else: #E2
            expt = 'E2'
        mouse = int(file_metadata[1].strip('m'))

        print('Processing mouse', expt, mouse)
    
        all_coverage_timecourses[(expt, mouse)] = get_coverage_timecourses(f'{mgx_timecourse_dir}/{file}', span_dictionary)

```

```python
import matplotlib.lines as mlines

# Create custom handler for stacked lines
class StackedLineHandler:
    def __init__(self, colors=['red', 'blue', 'green'], yoffsets=[0, -0.2, -0.4]):
        self.colors = colors
        self.yoffsets = yoffsets
        
    def legend_artist(self, legend, orig_handle, fontsize, handlebox):
        x0, y0 = handlebox.xdescent, handlebox.ydescent
        width = handlebox.width
        height = handlebox.height
        
        # Create the stacked lines
        lines = []
        for color, yoffset in zip(self.colors, self.yoffsets):
            l = mlines.Line2D(
                [x0, x0 + width], 
                [y0 + height/2 + yoffset, y0 + height/2 + yoffset],
                color=color,
                linewidth=0.5
            )
            lines.append(l)
            handlebox.add_artist(l)
            
        return lines
```
